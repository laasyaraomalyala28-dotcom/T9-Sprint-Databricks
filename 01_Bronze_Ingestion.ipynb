```python
%sql
LIST 's3://retail-sales-datawarehouse/sftp/';
```

```python
%sql

-- Step 1: Create schema
CREATE SCHEMA IF NOT EXISTS retail_etl;
```

```python
%sql
-- Step 2: Customers Bronze Table
DROP TABLE IF EXISTS retail_etl.bronze_customers;

CREATE TABLE retail_etl.bronze_customers
USING CSV
OPTIONS (
  path 's3a://retail-sales-datawarehouse/sftp/customers/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
-- Step 3: Products Bronze Table
DROP TABLE IF EXISTS retail_etl.bronze_products;

CREATE TABLE retail_etl.bronze_products
USING CSV
OPTIONS (
  path 's3://retail-sales-datawarehouse/sftp/products/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
-- Step 4: Stores Bronze Table
DROP TABLE IF EXISTS retail_etl.bronze_stores;

CREATE TABLE retail_etl.bronze_stores
USING CSV
OPTIONS (
  path 's3://retail-sales-datawarehouse/sftp/stores/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
-- Step 5: Sales Bronze Table

DROP TABLE IF EXISTS retail_etl.bronze_sales;

CREATE TABLE retail_etl.bronze_sales
USING CSV
OPTIONS (
  path 's3://retail-sales-datawarehouse/sftp/sales/',
  header 'true',
  inferSchema 'true'
);
```

```python
%sql
-- Step 6: Validation (Row Count Check)
SELECT 'customers' AS table_name, COUNT(*) AS cnt FROM retail_etl.bronze_customers
UNION ALL
SELECT 'products', COUNT(*) FROM retail_etl.bronze_products
UNION ALL
SELECT 'stores', COUNT(*) FROM retail_etl.bronze_stores
UNION ALL
SELECT 'sales', COUNT(*) FROM retail_etl.bronze_sales;
```
