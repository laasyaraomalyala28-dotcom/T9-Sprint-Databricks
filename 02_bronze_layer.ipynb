```python
%sql
-- CUSTOMERS BRONZE TABLE
CREATE OR REPLACE TABLE retail_lakehouse.bronze.customers
USING DELTA
AS
SELECT *
FROM read_files(
    's3a://retail-sales-datawarehouse/bronze/customers/',
    format => 'csv',
    header => true
);
```

```python
%sql
-- PRODUCTS BRONZE TABLE

CREATE OR REPLACE TABLE retail_lakehouse.bronze.products
USING DELTA
AS
SELECT *
FROM read_files(
    's3a://retail-sales-datawarehouse/bronze/products/',
    format => 'csv',
    header => true
);
```

```python
%sql
-- STORES BRONZE TABLE

CREATE OR REPLACE TABLE retail_lakehouse.bronze.stores
USING DELTA
AS
SELECT *
FROM read_files(
    's3a://retail-sales-datawarehouse/bronze/stores/',
    format => 'csv',
    header => true
);
```

```python
%sql
-- SALES BRONZE TABLE

CREATE OR REPLACE TABLE retail_lakehouse.bronze.sales
USING DELTA
AS
SELECT *
FROM read_files(
    's3a://retail-sales-datawarehouse/bronze/sales/',
    format => 'csv',
    header => true
);
```

```python
%sql
-- BRONZE VALIDATION

SELECT COUNT(*) AS customer_count
FROM retail_lakehouse.bronze.customers;

SELECT COUNT(*) AS product_count
FROM retail_lakehouse.bronze.products;

SELECT COUNT(*) AS store_count
FROM retail_lakehouse.bronze.stores;

SELECT COUNT(*) AS sales_count
FROM retail_lakehouse.bronze.sales;
```

```python
%sql
-- SAMPLE DATA VALIDATION

SELECT *
FROM retail_lakehouse.bronze.customers
LIMIT 5;

SELECT *
FROM retail_lakehouse.bronze.products
LIMIT 5;

SELECT *
FROM retail_lakehouse.bronze.stores
LIMIT 5;

SELECT *
FROM retail_lakehouse.bronze.sales
LIMIT 5;
```
